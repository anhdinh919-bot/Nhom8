```python
import sympy as sp

# ========================================================
# PHẦN 1: CÁC PHÉP TOÁN MA TRẬN CƠ BẢN
# ========================================================
def solve_matrices():
    print("=======================================")
    print(" PHẦN 1: CÁC PHÉP TOÁN MA TRẬN CƠ BẢN")
    print("=======================================\n")

    # Cộng hai ma trận
    A_add = sp.Matrix([[1, 4, -2], [3, 0, 5]])
    B_add = sp.Matrix([[-1, 2, 3], [4, -3, 1]])
    res_add = A_add + B_add
    print("Cộng ma trận:\n", res_add, "\n")

    # Trừ hai ma trận
    A_sub = sp.Matrix([[5, -1], [2, 7], [0, 4]])
    B_sub = sp.Matrix([[2, 3], [-1, 4], [1, 0]])
    res_sub = A_sub - B_sub
    print("Trừ ma trận:\n", res_sub, "\n")

    # Nhân vô hướng với ma trận
    A_scalar_1 = sp.Matrix([[2, -1, 0], [3, 5, -2]])
    res_scalar_1 = 4 * A_scalar_1  # SỬA LỖI: Thêm dấu *
    print("Nhân vô hướng 4:\n", res_scalar_1, "\n")

    # Nhân phân số với ma trận
    A_scalar_2 = sp.Matrix([[-6, 9], [3, 0], [-12, 15]])
    res_scalar_2 = -sp.Rational(1, 3) * A_scalar_2
    print("Nhân vô hướng -1/3:\n", res_scalar_2, "\n")

    # Tích hai ma trận vuông cấp 2
    A_mul_2x2 = sp.Matrix([[2, -1], [3, 4]])
    B_mul_2x2 = sp.Matrix([[1, 5], [-2, 0]])
    res_mul_2x2 = A_mul_2x2 * B_mul_2x2
    print("Tích 2 ma trận vuông:\n", res_mul_2x2, "\n")

    # Tích ma trận 2x3 và 3x2
    A_mul_rect = sp.Matrix([[1, -2, 3], [0, 4, -1]])
    B_mul_rect = sp.Matrix([[2, -1], [1, 3], [-2, 0]])
    res_mul_rect = A_mul_rect * B_mul_rect
    print("Tích ma trận 2x3 và 3x2:\n", res_mul_rect, "\n")

    # Ma trận chuyển vị (cấp 2x3)
    A_trans_1 = sp.Matrix([[3, -1, 5], [0, 2, -4]])
    res_trans_1 = A_trans_1.T
    print("Chuyển vị ma trận (2x3):\n", res_trans_1, "\n")

    # Ma trận chuyển vị (cấp 4x2)
    A_trans_2 = sp.Matrix([[1, 0], [-3, 7], [2, 5], [4, -1]])
    res_trans_2 = A_trans_2.T
    print("Chuyển vị ma trận (4x2):\n", res_trans_2, "\n")

    # Định thức ma trận cấp 3x3
    A_det_3x3 = sp.Matrix([[2, -1, 3], [4, 1, 5], [-2, 0, 1]])
    res_det_3x3 = A_det_3x3.det()
    print("Định thức cấp 3:\n", res_det_3x3, "\n")

    # Định thức ma trận cấp 4x4
    A_det_4x4 = sp.Matrix([[2, 1, 0, -1], [-1, 3, 2, 0], [0, 4, -2, 1], [3, 0, 1, -3]])
    res_det_4x4 = A_det_4x4.det()
    print("Định thức cấp 4:\n", res_det_4x4, "\n")

    # Ma trận nghịch đảo cấp 2x2
    A_inv_2x2 = sp.Matrix([[2, 5], [1, 3]])
    res_inv_2x2 = A_inv_2x2.inv()
    print("Ma trận nghịch đảo cấp 2:\n", res_inv_2x2, "\n")

    # Ma trận nghịch đảo cấp 3x3
    A_inv_3x3 = sp.Matrix([[1, 2, 0], [0, 1, 1], [1, 0, -1]])
    res_inv_3x3 = A_inv_3x3.inv()
    print("Ma trận nghịch đảo cấp 3:\n", res_inv_3x3, "\n")


# ========================================================
# PHẦN 2: BIẾN ĐỔI SƠ CẤP & HỆ PHƯƠNG TRÌNH TUYẾN TÍNH
# ========================================================
def matrix_operations():
    print("=======================================")
    print(" PHẦN 2: BIẾN ĐỔI VÀ HỆ PHƯƠNG TRÌNH")
    print("=======================================\n")

    # 1. KHAI BÁO MA TRẬN VÀ TRUY CẬP PHẦN TỬ
    print("--- 1. KHAI BÁO MA TRẬN ---")
    A_goc = sp.Matrix([
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]
    ])

    print("Ma trận A ban đầu:\n", A_goc)
    print("Phần tử a_11 (chỉ số [0, 0]):", A_goc[0, 0])

    # 2. CÁC PHÉP BIẾN ĐỔI SƠ CẤP TRÊN HÀNG
    print("\n--- 2. ĐỔI CHỖ HAI HÀNG (R1 <-> R3) ---")
    A1 = A_goc.copy()
    A1.row_swap(0, 2)
    print("Kết quả sau khi đổi chỗ R1 và R3:\n", A1)

    print("\n--- 3. NHÂN MỘT HÀNG VỚI MỘT SỐ KHÁC 0 (R2 -> 2*R2) ---")
    A2 = A_goc.copy()
    A2.row_op(1, lambda val, j: 2 * val)  # SỬA LỖI: 2*val thay vì 2val
    print("Kết quả sau khi nhân hàng 2 với 2:\n", A2)

    print("\n--- 4. CỘNG MỘT BỘI CỦA HÀNG VÀO HÀNG KHÁC (R2 -> R2 - 2*R1) ---")
    A3 = A_goc.copy()
    A3.row_op(1, lambda val, j: val - 2 * A_goc[0, j])  # SỬA LỖI: gộp dòng và thêm dấu *
    print("Kết quả sau khi biến đổi R2 -> R2 - 2*R1:\n", A3)

    # 3. DẠNG BẬC THANG VÀ DẠNG BẬC THANG RÚT GỌN (RREF)
    print("\n--- 5. DẠNG BẬC THANG VÀ BẬC THANG RÚT GỌN ---")
    A_bt = sp.Matrix([
        [1, 2, 3],
        [2, 4, 6],
        [1, 1, 2]
    ])

    E = A_bt.echelon_form()
    print("Dạng bậc thang (echelon_form):\n", E)

    R, pivot_columns = A_bt.rref()
    print("Dạng bậc thang rút gọn (RREF):\n", R)
    print("Các cột trụ (pivot columns):", pivot_columns)

    # 4. TÍNH HẠNG CỦA MA TRẬN
    print("\n--- 6. TÍNH HẠNG CỦA MA TRẬN ---")
    r = A_bt.rank()
    print("Hạng của ma trận (A_bt.rank()):", r)
    print("Hạng tính qua số cột trụ len(pivot_columns):", len(pivot_columns))

    # Ma trận đặc biệt
    O = sp.zeros(2, 2)
    I = sp.eye(3)
    print("Hạng ma trận không 2x2:", O.rank())
    print("Hạng ma trận đơn vị 3x3:", I.rank())

    # 5. GIẢI HỆ PHƯƠNG TRÌNH TUYẾN TÍNH
    print("\n--- 7. GIẢI HỆ PHƯƠNG TRÌNH (NGHIỆM DUY NHẤT) ---")
    A_sys = sp.Matrix([
        [1, 2, 1],
        [2, -1, 1],
        [1, 1, -1]
    ])
    B_sys = sp.Matrix([4, 1, 0])

    # Cách 1: Dùng ma trận mở rộng và RREF
    Ab = A_sys.row_join(B_sys)
    R_sys, pivots = Ab.rref()
    print("Ma trận mở rộng RREF:\n", R_sys)

    # Cách 2: Dùng gauss_jordan_solve()
    sol_gj, params = A_sys.gauss_jordan_solve(B_sys)
    print("Giải bằng gauss_jordan_solve():\n", sol_gj)

    # Cách 3: Dùng linsolve()
    sol_lin = sp.linsolve((A_sys, B_sys))
    print("Giải bằng linsolve():", sol_lin)

    # Kiểm tra định lý Rouché-Capelli
    print("rank(A) =", A_sys.rank())
    print("rank(Ab) =", Ab.rank())

    # 6. XÉT HỆ VÔ NGHIỆM VÀ VÔ SỐ NGHIỆM
    print("\n--- 8. HỆ VÔ NGHIỆM ---")
    A_vn = sp.Matrix([[1, 1], [2, 2]])
    B_vn = sp.Matrix([1, 3])
    Ab_vn = A_vn.row_join(B_vn)

    print("rank(A) =", A_vn.rank())
    print("rank(Ab) =", Ab_vn.rank())
    print("Tập nghiệm (linsolve):", sp.linsolve((A_vn, B_vn)))

    print("\n--- 9. HỆ VÔ SỐ NGHIỆM ---")
    A_vs = sp.Matrix([[1, 1], [2, 2]])
    B_vs = sp.Matrix([2, 4])
    Ab_vs = A_vs.row_join(B_vs)

    print("rank(A) =", A_vs.rank())
    print("rank(Ab) =", Ab_vs.rank())
    print("Tập nghiệm (linsolve):", sp.linsolve((A_vs, B_vs)))

    # 7. VÍ DỤ TỔNG HỢP
    print("\n--- 10. VÍ DỤ TỔNG HỢP ---")
    A_th = sp.Matrix([
        [1, 2, -1],
        [2, 4, -2],
        [1, 1, 1]
    ])
    B_th = sp.Matrix([2, 4, 3])
    Ab_th = A_th.row_join(B_th)

    print("Ma trận mở rộng:\n", Ab_th)
    print("Hạng của A:", A_th.rank())
    print("Hạng của ma trận mở rộng:", Ab_th.rank())

    R_th, _ = Ab_th.rref()
    print("Dạng bậc thang rút gọn:\n", R_th)
    print("Tập nghiệm:", sp.linsolve((A_th, B_th)))

    # 8. THỬ BÀI TẬP THỰC HÀNH (BÀI 1 & BÀI 3)
    print("\n--- 11. BÀI TẬP THỰC HÀNH (MẪU BÀI 3) ---")
    A_bt3 = sp.Matrix([
        [1, 1, 1],
        [2, -1, 1],
        [1, 2, -1]
    ])
    B_bt3 = sp.Matrix([6, 3, 2])
    print("Nghiệm bài tập 3:", sp.linsolve((A_bt3, B_bt3)))

# Chạy tất cả code
if __name__ == "__main__":
    solve_matrices()
    matrix_operations()
```
