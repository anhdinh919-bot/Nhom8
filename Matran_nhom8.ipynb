```python
import sympy as sp

# Khởi tạo hiển thị đẹp nếu chạy trên Jupyter Notebook
sp.init_printing()

# ========================================================
# 1. KHAI BÁO MA TRẬN VÀ TRUY CẬP PHẦN TỬ
# ========================================================
print("--- 1. KHAI BÁO MA TRẬN ---")
A_goc = sp.Matrix([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
])

print("Ma trận A ban đầu:")
print(A_goc)
print("Phần tử a_11 (chỉ số [0, 0]):", A_goc[0, 0])


# ========================================================
# 2. CÁC PHÉP BIẾN ĐỔI SƠ CẤP TRÊN HÀNG
# (Dùng .copy() để tránh lỗi sửa đổi trực tiếp lên A_goc)
# ========================================================
print("\n--- 2. ĐỔI CHỖ HAI HÀNG (R1 <-> R3) ---")
A1 = A_goc.copy()
A1.row_swap(0, 2)
print("Kết quả sau khi đổi chỗ R1 và R3:")
print(A1)


print("\n--- 3. NHÂN MỘT HÀNG VỚI MỘT SỐ KHÁC 0 (R2 -> 2*R2) ---")
A2 = A_goc.copy()
A2.row_op(1, lambda val, j: 2 * val)
print("Kết quả sau khi nhân hàng 2 với 2:")
print(A2)


print("\n--- 4. CỘNG MỘT BỘI CỦA HÀNG VÀO HÀNG KHÁC (R2 -> R2 - 2*R1) ---")
A3 = A_goc.copy()
# Lưu ý: Lấy giá trị hàng 1 từ A_goc ban đầu để đảm bảo tính chính xác
A3.row_op(1, lambda val, j: val - 2 * A_goc[0, j])
print("Kết quả sau khi biến đổi R2 -> R2 - 2*R1:")
print(A3)


# ========================================================
# 3. DẠNG BẬC THANG VÀ DẠNG BẬC THANG RÚT GỌN (RREF)
# ========================================================
print("\n--- 5. DẠNG BẬC THANG VÀ BẬC THANG RÚT GỌN ---")
A_bt = sp.Matrix([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 2]
])

E = A_bt.echelon_form()
print("Dạng bậc thang (echelon_form):")
print(E)

R, pivot_columns = A_bt.rref()
print("Dạng bậc thang rút gọn (RREF):")
print(R)
print("Các cột trụ (pivot columns):", pivot_columns)


# ========================================================
# 4. TÍNH HẠNG CỦA MA TRẬN
# ========================================================
print("\n--- 6. TÍNH HẠNG CỦA MA TRẬN ---")
r = A_bt.rank()
print("Hạng của ma trận (A_bt.rank()):", r)
print("Hạng tính qua số cột trụ len(pivot_columns):", len(pivot_columns))

# Ma trận đặc biệt
O = sp.zeros(2, 2)
I = sp.eye(3)
print("Hạng ma trận không 2x2:", O.rank())
print("Hạng ma trận đơn vị 3x3:", I.rank())


# ========================================================
# 5. GIẢI HỆ PHƯƠNG TRÌNH TUYẾN TÍNH
# ========================================================
print("\n--- 7. GIẢI HỆ PHƯƠNG TRÌNH (NGHIỆM DUY NHẤT) ---")
A_sys = sp.Matrix([
    [1, 2, 1],
    [2, -1, 1],
    [1, 1, -1]
])

B_sys = sp.Matrix([4, 1, 0])

# Cách 1: Dùng ma trận mở rộng và RREF
Ab = A_sys.row_join(B_sys)
R_sys, pivots = Ab.rref()
print("Ma trận mở rộng RREF:")
print(R_sys)

# Cách 2: Dùng gauss_jordan_solve()
sol_gj, params = A_sys.gauss_jordan_solve(B_sys)
print("Giải bằng gauss_jordan_solve():")
print(sol_gj)

# Cách 3: Dùng linsolve()
sol_lin = sp.linsolve((A_sys, B_sys))
print("Giải bằng linsolve():", sol_lin)

# Kiểm tra định lý Rouché-Capelli
print("rank(A) =", A_sys.rank())
print("rank(Ab) =", Ab.rank())


# ========================================================
# 6. XÉT HỆ VÔ NGHIỆM VÀ VÔ SỐ NGHIỆM
# ========================================================
print("\n--- 8. HỆ VÔ NGHIỆM ---")
A_vn = sp.Matrix([[1, 1], [2, 2]])
B_vn = sp.Matrix([1, 3])
Ab_vn = A_vn.row_join(B_vn)

print("rank(A) =", A_vn.rank())
print("rank(Ab) =", Ab_vn.rank())
print("Tập nghiệm (linsolve):", sp.linsolve((A_vn, B_vn)))


print("\n--- 9. HỆ VÔ SỐ NGHIỆM ---")
A_vs = sp.Matrix([[1, 1], [2, 2]])
B_vs = sp.Matrix([2, 4])
Ab_vs = A_vs.row_join(B_vs)

print("rank(A) =", A_vs.rank())
print("rank(Ab) =", Ab_vs.rank())
print("Tập nghiệm (linsolve):", sp.linsolve((A_vs, B_vs)))


# ========================================================
# 7. VÍ DỤ TỔNG HỢP
# ========================================================
print("\n--- 10. VÍ DỤ TỔNG HỢP ---")
A_th = sp.Matrix([
    [1, 2, -1],
    [2, 4, -2],
    [1, 1, 1]
])
B_th = sp.Matrix([2, 4, 3])
Ab_th = A_th.row_join(B_th)

print("Ma trận mở rộng:")
print(Ab_th)
print("Hạng của A:", A_th.rank())
print("Hạng của ma trận mở rộng:", Ab_th.rank())

R_th, _ = Ab_th.rref()
print("Dạng bậc thang rút gọn:")
print(R_th)
print("Tập nghiệm:", sp.linsolve((A_th, B_th)))


# ========================================================
# 8. THỬ BÀI TẬP THỰC HÀNH (BÀI 1 & BÀI 3)
# ========================================================
print("\n--- 11. BÀI TẬP THỰC HÀNH (MẪU BÀI 3) ---")
A_bt3 = sp.Matrix([
    [1, 1, 1],
    [2, -1, 1],
    [1, 2, -1]
])
B_bt3 = sp.Matrix([6, 3, 2])
print("Nghiệm bài tập 3:", sp.linsolve((A_bt3, B_bt3)))
```
